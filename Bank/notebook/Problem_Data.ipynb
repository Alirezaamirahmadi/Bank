```python
# وارد کردن کتابخانه‌ها
import numpy as np
import pandas as pd

# خواندن داده 
df = pd.read_csv("bank-full.csv", sep=";")

# نمایش پنج سطر اول
df.head()
```

```python
# نمایش تعداد سطرها و ستون‌ها

print(f"Number os Records: {df.shape[0]:,}")
print(f"Number os Columns: {df.shape[1]:,}")
```

```python
# نمایش اطلاعات کلی

df.info()
```

```python
# نمایش آمار توصیفی تمام ستون‌ها

df.describe(include="all").T
```

```python
# مشخص کردن نام ستون هدف
TARGET = "y"

# جدا کردن
X = df.drop(columns=TARGET)

# ذخیره Target در متغیر y
y = df[TARGET]

# نمایش ابعاد
print("Scale of Features:", X.shape)
print("Scale of Target:", y.shape)

# نمایش نام تمام Feature
print("\nList of Features:")
print(X.columns.tolist())
```

```python
# شناسایی Featureهای عددی
numeric_features = X.select_dtypes(include=np.number).columns.tolist()

# شناسایی Featureهای categorical
categorical_features = X.select_dtypes(include="object").columns.tolist()

# نمایش Featureهای عددی
print("Numeric Features:")
print(numeric_features)

# categorical Features
print("\ncategorical Features:")
print(categorical_features)
```

```python
# محاسبه تعداد نمونه‌های هر کلاس
target_counts = y.value_counts()

# محاسبه درصد هر کلاس
target_percentages = y.value_counts(normalize=True) * 100

# ساخت یک جدول خلاصه برای بررسی توزیع
target_summary = pd.DataFrame({
    "Count": target_counts,
    "Percent": target_percentages.round(2)
})

# نمایش توزیع 
target_summary
```

```python
# محاسبه تعداد مقادیر خالی
missing_values = df.isnull().sum().sort_values(ascending=False)

# Missing Values
missing_values[missing_values > 0]
```

```python
# شمارش رکوردهای تکراری
duplicate_count = df.duplicated().sum()

print(f"Duplicates: {duplicate_count:,}")
```

```python
# ساخت یک جدول برای بررسی نوع داده و تعداد مقادیر یکتا در هر ستون

feature_info = pd.DataFrame({
    "feature": df.columns,
    "dtype": df.dtypes.astype(str),
    "unique_values": df.nunique().values
})

# نمایش جدول اطلاعات Featureها
feature_info
```

```python
# categorical Features

for column in categorical_features:

    print(f"\n{'=' * 60}")
    print(f"Column: {column}")

    # نمایش پرتکرارترین مقادیر این ستون
    print(df[column].value_counts(dropna=False).head(15))
```

```python
# نمایش آمار توصیفی Featureهای عددی
# این بخش برای بررسی اولیه دامنه مقادیر و Outlierهای احتمالی مفید است

df[numeric_features].describe().T
```

```python
# محاسبه تعداد نمونه‌های هر کلاس
class_counts = y.value_counts()

# محاسبه درصد هر کلاس
class_percentages = y.value_counts(normalize=True) * 100

# ساخت جدول نهایی
class_distribution = pd.DataFrame({
    "Counts": class_counts,
    "Percent": class_percentages.round(2)
})

# نمایش توزیع کلاس‌ها
class_distribution
```

```python
# بررسی تعداد مقادیر یکتا

cardinality = pd.DataFrame({
    "feature": X.columns,
    "unique_values": X.nunique(),
    "dtype": X.dtypes.astype(str)
})

# مرتب‌سازی بر اساس تعداد مقادیر یکتا
cardinality.sort_values("unique_values", ascending=False)
```

```python
# بررسی حداقل، حداکثر، میانگین

numerical_summary = X[numeric_features].describe(
    percentiles=[0.01, 0.25, 0.50, 0.75, 0.99]
).T

numerical_summary
```

```python
# نمایش کمترین و بیشترین مقدار

feature_ranges = pd.DataFrame({
    "Min": X[numeric_features].min(),
    "Max": X[numeric_features].max(),
    "Mean": X[numeric_features].mean(),
    "Median": X[numeric_features].median()
})

feature_ranges
```

```python
# بررسی نرخ پذیرش سپرده برای هر گروه

categorical_for_eda = [
    "job",
    "marital",
    "education",
    "default",
    "housing",
    "loan",
    "contact",
    "month",
    "poutcome"
]

for column in categorical_for_eda:

    print(f"\n{'=' * 70}")
    print(f"Feature: {column}")

    # محاسبه نرخ yes برای هر دسته
    acceptance_rate = (
        df.groupby(column)["y"]
        .apply(lambda values: (values == "yes").mean() * 100)
        .sort_values(ascending=False)
    )

    print(acceptance_rate.round(2))
```

```python
# Feature مربوط به مدت تماس را از ورودی مدل حذف می‌کنیم
# دلیل: این مقدار بعد از تماس مشخص می‌شود و در سناریوی Prediction
# قبل از تماس در دسترس نیست

LEAKAGE_FEATURES = ["duration"]

# ساخت Featureهای نهایی برای مدل
X_model = X.drop(columns=LEAKAGE_FEATURES)

# Target بدون تغییر باقی می‌ماند
y_model = y.copy()

# نمایش Featureهای نهایی
print("Features used by the model:")
print(X_model.columns.tolist())

print(f"\nFinal Count of Features: {X_model.shape[1]}")
```

```python
# شناسایی Featureهای نهایی
numeric_features_model = (
    X_model
    .select_dtypes(include=np.number)
    .columns
    .tolist()
)

# شناسایی Featureهای categorical نهایی
categorical_features_model = (
    X_model
    .select_dtypes(include="object")
    .columns
    .tolist()
)

print("Numeric Features:")
print(numeric_features_model)

print("\ncategorical Features:")
print(categorical_features_model)
```
